**Memory Management Function**

```python
def optimize_memory(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            df[col] = df[col].astype('category')
        elif df[col].dtype == 'float64':
            df[col] = df[col].astype('float32')
        elif df[col].dtype == 'int64':
            df[col] = pd.to_numeric(df[col], downcast='integer')
    return df

# Apply optimization
train = optimize_memory(train)
test = optimize_memory(test)

print("\nMemory optimization completed.")
train.info(memory_usage='deep')
```

#2. Data Cleaning

**Handling Missing Values**

```python
# check for missing values

missing_values = train.isnull().sum()

# Keep only columns with missing values
missing_values = missing_values[missing_values > 0]

# Sort values
missing_values.sort_values(ascending=False)
```

```python
# Plot missing values
plt.figure(figsize=(8, 6))
missing_values.head(20).plot(kind="bar", figsize=(8, 6))

plt.xlabel("Columns", fontsize=20)
plt.ylabel("Count", fontsize=20)
plt.title("Total Missing Values", fontsize=20)
plt.show()
```

### Rationale for Missing Values Imputation:
We filled numerical missing values using the **median** rather than the mean. The median is a robust statistic that is not influenced by extreme outliers, ensuring a more accurate and stable imputation for the skewed features in our dataset.

```python
# Fill special categorical features

# الفئات الخاصة المطلوبة في الواجب
special_columns = [
    "PoolQC", "MiscFeature", "Alley", "Fence", "FireplaceQu",
    "GarageType", "GarageFinish", "GarageQual", "GarageCond",
    "BsmtExposure", "BsmtFinType1", "BsmtFinType2", "BsmtQual", "BsmtCond"
]

for column in special_columns:
    # تعديل بيانات التدريب
    if column in train.columns:
        train[column] = train[column].astype(str).replace('nan', 'None').fillna('None').astype('category')

    # تعديل بيانات الاختبار
    if column in test.columns:
        test[column] = test[column].astype(str).replace('nan', 'None').fillna('None').astype('category')
```

```python

# Fill numerical columns using median

numeric_columns = train.select_dtypes(include=[np.number]).columns

for column in numeric_columns:

    train[column] = train[column].fillna(train[column].median())

    # Skip SalePrice because it does not exist in test dataset

    if column != "SalePrice":

        test[column] = test[column].fillna(test[column].median())
```

```python

# Fill remaining categorical columns using mode

categorical_columns = train.select_dtypes(include=["object"]).columns

for column in categorical_columns:

    train[column] = train[column].fillna(train[column].mode()[0])

    test[column] = test[column].fillna(test[column].mode()[0])
```

#4. Feature Engineering

```python
# Create TotalSF feature

train["TotalSF"] = (
    train["TotalBsmtSF"]
    + train["1stFlrSF"]
    + train["2ndFlrSF"]
)

test["TotalSF"] = (
    test["TotalBsmtSF"]
    + test["1stFlrSF"]
    + test["2ndFlrSF"]
)
```

**Encoding Categorical Variables**

```python
# Label Encoding for Ordinal Features (Preserving Rank)
# Define the quality mapping based on the dataset description
qual_map = {'Ex': 5, 'Gd': 4, 'TA': 3, 'Fa': 2, 'Po': 1, 'None': 0}
ordinal_cols = ['ExterQual', 'KitchenQual', 'BsmtQual', 'HeatingQC']

for col in ordinal_cols:
    if col in train.columns:
        # Convert to string to handle category dtypes and apply mapping
        train[col] = train[col].astype(str).map(qual_map).fillna(0)
    if col in test.columns:
        test[col] = test[col].astype(str).map(qual_map).fillna(0)
```

```python
# One-Hot Encoding for Nominal Features
# We encode 'Neighborhood' and 'BldgType' as per requirements
train = pd.get_dummies(train, columns=['Neighborhood', 'BldgType'], drop_first=True)
test = pd.get_dummies(test, columns=['Neighborhood', 'BldgType'], drop_first=True)

print("Categorical variables encoded successfully.")
```

**Pipeline Construction**

```python


def final_preprocessing_pipeline(df):
    df_out = df.copy()

    # 1. Fill numeric missing values
    numeric_cols = df_out.select_dtypes(include=[np.number]).columns
    df_out[numeric_cols] = df_out[numeric_cols].fillna(df_out[numeric_cols].median())

    # 2. Add TotalSF if not already created
    if 'TotalSF' not in df_out.columns:
        df_out['TotalSF'] = df_out['TotalBsmtSF'].fillna(0) + df_out['1stFlrSF'] + df_out['2ndFlrSF']

    # 3. One-Hot Encoding inside the pipeline
    df_out = pd.get_dummies(df_out, columns=['Neighborhood', 'BldgType'], drop_first=True)

    return df_out
```
